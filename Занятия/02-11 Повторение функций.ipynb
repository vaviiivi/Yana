```python
# У нас есть диапазон значений от 1 до 1000. Нужно вывести все простые числа.
# 12 -> 1 2 3 4 6 12

def is_simple(x):
    count = 0
    for div in range(1, x + 1):
        if x % div == 0:
            count += 1
    return count == 2

for x in range(1, 1001):
    if is_simple(x):
        print(x)
```

```python
l = ["abc", "a", "babbbb", "cd"]

len(l)  # длина
l[0]  # нулевой элемент списка
l[2:3]  # срез

for x in l:
    print(x)
```

```python
# Вывести имена, начинающиеся на А
names = ["Kate", "Kris", "Yana", "Alex", "Alisa"]

for name in names:
    if name[0] == "A":
        print(name)
```
