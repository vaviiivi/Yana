Дан список values = [3, -8, 19, -12, 6, 9]. Заменить все отрицательные числа на нули.

```python
values = [3, -8, 19, -12, 6, 9]
for index in range(0, len(values)):
    if values[index] < 0:
        values[index] = 0
print(values)
```

Дан список values = [3, -8, 19, -12, 6, 9]. Найти максимальное значение.

```python
values = [3, -8, 19, -12, 6, 9]
maxi = -100
for value in values:
    if value > maxi:
        maxi = value
print(maxi)
```
