Дан список строк. Нужно каждую строку с четным номером в этом списке заменить на "".

```python
s = ["abc", "ab", "cc", "abcd", "acd"]
for i in range(len(s)):
    if (i + 1) % 2 == 0:
        s[i] = ""
print(s)
```

Дан список строк. Нужно каждую строку с четным номером в этом списке заменить на "". И в новый список добавить остальные непустые строки.

```python
s = ["abc", "ab", "cc", "abcd", "acd"]
y=[]
for i in range(0,len(s)):
    if (i+1) % 2 != 0:
        y.append(s[i])
print(y)
```

Дан список строк. Найти строку с максимальной длиной и поставить на ее место "".

```python
s = ["abc", "ab", "cc", "abcd", "acd"]
maxx = 0
maxx_i = 0
for i in range(len(s)):
    if len(s[i]) > maxx:
        maxx = len(s[i])
        maxx_i = i
s[maxx_i] = ""
print(s)
```
